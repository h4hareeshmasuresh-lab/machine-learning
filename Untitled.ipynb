# Statistics and Probability for Data Science
## 1. Descriptive Statistics
Descriptive statistics are used to summarize and understand a dataset. The main measures are Mean, Median, Mode, Variance, and Standard Deviation.

###  Mean (Average)
The arithmetic mean is the sum of all observations divided by the total number of observations.

#### Formula:
Mean = Sum of all observations / Total number of observations

### Example:

Data = [10, 20, 30, 40, 50]

Mean = (10 + 20 + 30 + 40 + 50) / 5

Mean = 150 / 5 = 30

#### Python Code:

```python
import statistics as stats

data = [10, 20, 30, 40, 50]

print("Mean:", stats.mean(data))
```

## Median
The median is the middle value of a dataset when the data is arranged in ascending or descending order.

If the number of observations is odd, the median is the middle value.
If the number of observations is even, the median is the average of the two middle values.
### Example 1: Odd number of observations

Data = [10, 20, 30, 40, 50]

Median = 30

### Example 2: Even number of observations

Data = [10, 20, 30, 40]

Median = (20 + 30) / 2 = 25

#### Python Code:

```python
import statistics as stats

data = [10, 20, 30, 40, 50]

print("Median:", stats.median(data))
```

## Mode
The mode is the value that appears most frequently in a dataset.

A dataset can have:
+ Unimodal: One mode.
+ Bimodal: Two modes.
+ Multimodal: More than two modes.
+ No mode: No value occurs more frequently than the others.
### Example:

Data = [10, 20, 20, 30, 40]

Mode = 20

#### Python Code:

```python
import statistics as stats

data = [10, 20, 20, 30, 40]

print("Mode:", stats.mode(data))
```

### Variance
Variance measures how spread out the data points are from the mean.

A higher variance indicates that the data points are more spread out, while a lower variance indicates that they are closer to the mean.

Sample Variance Formula:

Variance = Σ(x - Mean)² / (n - 1)

Where:

+ x = Each observation
+ Mean = Average of the observations
+ n = Total number of observations
### Example:

Data = [10, 20, 30, 40, 50]

Mean = 30

Variance = [(10-30)² + (20-30)² + (30-30)² + (40-30)² + (50-30)²] / (5-1)

Variance = (400 + 100 + 0 + 100 + 400) / 4

Variance = 1000 / 4

Variance = 250

#### Python Code:

```python
import statistics as stats

data = [10, 20, 30, 40, 50]

print("Variance:", stats.variance(data))
```

###  Standard Deviation
Standard deviation is the square root of variance. It measures how much the data points typically deviate from the mean.

Formula:

Standard Deviation = √Variance

### Example:

Variance = 250

Standard Deviation = √250

Standard Deviation = 15.811

#### Python Code:

```python
import statistics as stats

data = [10, 20, 30, 40, 50]

print("Standard Deviation:", stats.stdev(data))
```

## 2. Probability Basics
Probability is one of the most important topics in Statistics and Data Science.

Probability measures how likely an event is to occur.

The value of probability is always between 0 and 1.

Formula:

P(E) = Number of favourable outcomes / Total number of possible outcomes

Probability Range:

0 ≤ P(E) ≤ 1

+ P(E) = 0: The event will never happen.
+ P(E) = 1: The event will definitely happen.
+ P(E) = 0.5: The event has an equal chance of happening or not happening.
### Example: Tossing a Coin

Total outcomes = 2 (Head, Tail)

Favourable outcome for getting Head = 1

P(Head) = 1 / 2 = 0.5



## 3. Important Probability Rules
### Complementary Rule
The complementary rule gives the probability that an event will NOT happen.

#### Formula:

P(Not A) = 1 - P(A)

### Example:

If P(A) = 0.7

P(Not A) = 1 - 0.7

P(Not A) = 0.3

### Addition Rule (OR Events)
The addition rule calculates the probability of event A or event B occurring.

#### Formula:

P(A or B) = P(A) + P(B) - P(A and B)

If two events are mutually exclusive (they cannot happen together):

P(A or B) = P(A) + P(B)

### Example: Rolling a Dice

Event A = Getting an even number (2, 4, 6)

P(A) = 3/6

Event B = Getting a number less than 4 (1, 2, 3)

P(B) = 3/6

Intersection = Getting an even number less than 4 (2)

P(A and B) = 1/6

Therefore:

P(A or B) = 3/6 + 3/6 - 1/6

P(A or B) = 5/6

###  Multiplication Rule (AND Events)
For two independent events, the multiplication rule calculates the probability of both events occurring.

#### Formula:

P(A and B) = P(A) × P(B)

### Example: Tossing Two Coins

P(Head on the first coin) = 1/2

P(Head on the second coin) = 1/2

P(Both Heads) = 1/2 × 1/2

P(Both Heads) = 1/4 = 0.25

##  Probability Examples Using Python
###  Tossing a Coin

```python
favourable = 1
total = 2

p_head = favourable / total

print("P(Head) =", p_head)
```

###  Rolling a Dice (Even Number)

```python
favourable = 3
total = 6

p_even = favourable / total

print("P(Even) =", p_even)
```

###  Complementary Probability

```python
p_not_even = 1 - p_even

print("P(Not Even) =", p_not_even)
```

# Types of Probability
### Theoretical Probability
Theoretical probability is based on mathematical reasoning rather than actual experiments.

### Example:

The probability of rolling a 6 on a fair dice is:

P(6) = 1/6

### Experimental Probability
Experimental probability is based on actual experiments or observations.

### Example:

A coin is tossed 100 times.

Number of times Heads appear = 48

Total number of tosses = 100

Experimental Probability = 48/100

Experimental Probability = 0.48

### Subjective Probability
Subjective probability is based on personal opinion, intuition, or experience.

### Example:

"I feel there is a 70% chance that it will rain today."

# PROBABILITY DISTRIBUTIONS
Probability distributions describe how probabilities are assigned to different possible outcomes of a random variable.

The main types of probability distributions are:
+ Normal Distribution
+ Binomial Distribution
+ Poisson Distribution
+ Uniform Distribution

## 1. Normal Distribution
The Normal Distribution is also called the Gaussian Distribution. It is one of the most common continuous probability distributions.

It is a bell-shaped curve that is symmetric around the mean.

### Probability Density Function (PDF):
f(x) = 1 / (σ√(2π)) × e^[-(x - μ)² / (2σ²)]
 



Where:
+ μ = Mean
+ σ = Standard deviation
+ x = Random variable
+ σ² = Variance

#### Python Program:

```python
import numpy as np
import matplotlib.pyplot as plt

# Mean and standard deviation
mu = 4
sigma = 1

# Generate random values
x = np.random.normal(mu, sigma, 1000)

# Plot histogram
plt.hist(x, bins=30, density=True, color='blue')

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Probability Density")

plt.show()
```

## 2. Binomial Distribution
The Binomial Distribution is used for discrete outcomes when there are only two possible results, such as Success or Failure.

### Examples:
+ Tossing a coin (Head or Tail).
+ Passing or failing an exam.
+ Yes/No type experiments.

### Probability Mass Function (PMF):
P(X = k) = nck × p^k × (1-p)^(n-k)

Where:
+ n = Number of trials
+ k = Number of successes
+ p = Probability of success
+ 1−p = Probability of failure

Mean and Variance:
+ Mean=np
+ Variance=np(1−p)
### Python Program:

```python
from scipy.stats import binom
import matplotlib.pyplot as plt

# Number of trials
n = 10

# Probability of success
p = 0.5

# Possible number of successes
x = range(0, n+1)

# Plot binomial distribution
plt.bar(x, binom.pmf(x, n, p), color='orange')

plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")

plt.show()
```

## 3. Poisson Distribution
The Poisson Distribution is used for counting the number of events that occur within a fixed interval of time or space.

### Examples:
+ Number of emails received per hour.
+ Number of accidents per day.
+ Number of calls arriving at a call centre per minute.

### Probability Mass Function (PMF):
P(X = k) = (e^(-λ) × λ^k) / k!

Where:
+ λ = Average number of events (mean).
+ k = Number of occurrences.
+ e = Euler's number (approximately 2.718).

#### Mean and Variance:
+ Mean=λ
+ Variance=λ
### Python Program:

```python
from scipy.stats import poisson
import matplotlib.pyplot as plt

# Average number of events
lam = 4

# Possible number of events
x = range(0, 10)

# Plot Poisson distribution
plt.bar(x, poisson.pmf(x, lam), color='green')

plt.title("Poisson Distribution (lambda=4)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")

plt.show()
```
